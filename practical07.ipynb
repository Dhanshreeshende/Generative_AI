```python

# Step 1: Install required libraries
!pip install transformers torch -q

# Step 2: Import libraries
import torch
from transformers import AutoTokenizer, AutoModelForQuestionAnswering

# Step 3: Load pre-trained BERT model
model_name = "distilbert-base-cased-distilled-squad"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

# Set model to evaluation mode
model.eval()

# Step 4: Define college regulations context
context = """
A student must have a minimum attendance of 75 percent in
each subject to be eligible for the end semester examination.
Students with attendance between 65 and 75 percent may apply
for condonation with a medical certificate.

The internal assessment carries 40 marks and the end semester
examination carries 60 marks. A student must score at least
40 percent overall to pass a course.

The library remains open from 9 AM to 6 PM on all working days.
Fee payment must be completed before the last date announced
by the accounts section, and a late fee of 500 rupees is charged
after the due date.

Scholarship applications are submitted through the college
office within the first month of the semester.

Students must carry their college identity card while entering
the campus. The college canteen is open from 8 AM to 5 PM on
working days.

Practical examinations are conducted according to the
department schedule. Students must submit assignments before
the deadline given by their faculty.

The minimum passing marks in the end semester examination
are 24 out of 60.

Students can borrow up to three books from the library at a time.
A library book can be borrowed for 14 days.

Students must contact their department office for
timetable-related queries.

The college provides sports facilities for students after
regular lecture hours.
"""

print("College regulations loaded successfully.")

# Step 5: Define the question-answering function
def answer(question):
    inputs = tokenizer(
        question,
        context,
        return_tensors="pt",
        truncation="only_second",
        max_length=384
    )

    with torch.no_grad():
        outputs = model(**inputs)

    # Get predicted start and end positions
    start = int(outputs.start_logits.argmax())
    end = int(outputs.end_logits.argmax())

    # Check whether the predicted span is valid
    if end < start:
        return "Answer not found"

    # Extract and decode the answer
    answer_ids = inputs["input_ids"][0][start:end + 1]

    predicted_answer = tokenizer.decode(
        answer_ids,
        skip_special_tokens=True
    )

    return predicted_answer if predicted_answer.strip() else "Answer not found"


# Step 6: Test student questions
questions = [
    "What is the minimum attendance required for the exam?",
    "How many marks does the internal assessment carry?",
    "What is the late fee after the due date?",
    "When is the library open?",
    "What is the minimum overall percentage required to pass?",
    "When should scholarship applications be submitted?",
    "What should students carry when entering the campus?",
    "What are the college canteen timings?",
    "How many marks are required to pass the end semester examination?",
    "How many books can students borrow from the library?",
    "For how many days can a library book be borrowed?",
    "When are practical examinations conducted?",
    "When should students submit their assignments?",
    "Whom should students contact for timetable-related queries?",
    "When can students use sports facilities?"
]

# Step 7: Display predicted answers
for q in questions:
    print("Q:", q)
    print("A:", answer(q))
    print("-" * 50)
```
